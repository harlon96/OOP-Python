```python
def firstelt(L):    #defining a function
    return L[0]

Firstelt = lambda L: L[0]   #lambda anonymous function

print(firstelt(squares))
print(firstelt(squares))
```

```python
def replacefirst(L,v):  #concatenate .. linking things together
    return [v] + L[1:]  # [1:] is subscripting in Python

Replacefirst = lambda L, v: [v] + L[1:] #concatenate
```

```python
print(replacefirst(squares,7))
print(Replacefirst(squares,7))
```

```python
def replacefirst_mut(L,v): #no return ususally means mutating
    L[0] = v
```
